Null Values with numerical value

```python
#Replace null values with mean numerical value
healthcare['patient_id'] = healthcare['patient_id'].fillna(healthcare['patient_id'].mean())
healthcare['age'] = healthcare['age'].fillna(healthcare['age'].mean())
healthcare['treatment_cost'] = healthcare['treatment_cost'].fillna(healthcare['treatment_cost'].mean())
healthcare['insurance_coverage'] = healthcare['insurance_coverage'].fillna(healthcare['insurance_coverage'].mean())
healthcare['doctor_id'] = healthcare['doctor_id'].fillna(healthcare['doctor_id'].mean())
healthcare['readmission_flag'] = healthcare['readmission_flag'].fillna(healthcare['readmission_flag'].mean())
healthcare['length_of_stay'] = healthcare['length_of_stay'].fillna(healthcare['length_of_stay'].mean())
healthcare['medication_cost'] = healthcare['medication_cost'].fillna(healthcare['medication_cost'].mean())
healthcare['doctor_rating'] = healthcare['doctor_rating'].fillna(healthcare['doctor_rating'].mean())
healthcare['hospital_rating'] = healthcare['hospital_rating'].fillna(healthcare['hospital_rating'].mean())
```

```python
print(healthcare.isnull().sum())
```

null values with mode for categorical columns

```python
# Replace null values with mode for categorical columns
healthcare['gender'] = healthcare['gender'].fillna(healthcare['gender'].mode()[0])
healthcare['disease'] = healthcare['disease'].fillna(healthcare['disease'].mode()[0])
healthcare['hospital'] = healthcare['hospital'].fillna(healthcare['hospital'].mode()[0])
healthcare['admission_date'] = healthcare['admission_date'].fillna(healthcare['admission_date'].mode()[0])
healthcare['discharge_date'] = healthcare['discharge_date'].fillna(healthcare['discharge_date'].mode()[0])
healthcare['bed_type'] = healthcare['bed_type'].fillna(healthcare['bed_type'].mode()[0])
healthcare['city'] = healthcare['city'].fillna(healthcare['city'].mode()[0])
healthcare['state'] = healthcare['state'].fillna(healthcare['state'].mode()[0])
healthcare['test_results'] = healthcare['test_results'].fillna(healthcare['test_results'].mode()[0])
healthcare['severity_level'] = healthcare['severity_level'].fillna(healthcare['severity_level'].mode()[0])
```

```python
print(healthcare.isnull().sum())
```

Exploratory Data Analysis (EDA)

#Patient ages

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=healthcare, x="age", bins=20, kde=True)
plt.title("Distribution of Patient Ages")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

#Patients by disease

```python
plt.figure(figsize=(8, 5))
healthcare["disease"].value_counts().plot(kind="bar", color="teal")
plt.title("Number of Patients by Disease")
plt.xlabel("Disease")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#Average treatment cost by disease

```python
plt.figure(figsize=(8, 5))
healthcare.groupby("disease")["treatment_cost"].mean().sort_values().plot(
    kind="barh", color="grey"
)
plt.title("Average Treatment Cost by Disease")
plt.xlabel("Average Treatment Cost")
plt.ylabel("Disease")
plt.tight_layout()
plt.show()
```

#Readmissions by severity

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=healthcare, x="severity_level", hue="readmission_flag")
plt.title("Readmissions by Severity Level")
plt.xlabel("Severity Level")
plt.ylabel("Number of Patients")
plt.legend(title="Readmitted")
plt.show()
```

#Admissions by month

```python
healthcare["admission_date"] = pd.to_datetime(
    healthcare["admission_date"], errors="coerce"
)

monthly_admissions = (
    healthcare.dropna(subset=["admission_date"])
    .groupby(healthcare["admission_date"].dt.to_period("M"))
    .size()
)

monthly_admissions.plot(figsize=(10, 5), marker="o", color="purple")
plt.title("Monthly Patient Admissions")
plt.xlabel("Month")
plt.ylabel("Number of Admissions")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#Patients by insurance coverage

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=healthcare, x="insurance_coverage", color="skyblue")
plt.title("Patients by Insurance Coverage")
plt.xlabel("Insurance Coverage")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

#Patients by gender

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=healthcare, x="gender", color="lightgreen")
plt.title("Number of Patients by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7, 5))
sns.countplot(data=healthcare, x="hospital_rating", color="mediumpurple")
plt.title("Distribution of Hospital Ratings")
plt.xlabel("Hospital Rating")
plt.ylabel("Number of Patients")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
